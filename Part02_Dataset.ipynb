```python
import os
import sys

PROJECT_ROOT = os.path.abspath("..")

if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)
```

```python
import torch
from torchvision.datasets import Flowers102
```

```python
#download dataset
train_dataset = Flowers102(
    root="./dataset",
    split="train",
    download=True
)

val_dataset = Flowers102(
    root="./dataset",
    split="val",
    download=True
)

test_dataset = Flowers102(
    root="./dataset",
    split="test",
    download=True
)
```

```python
#Check the dataset size
print("Training Images :", len(train_dataset))
print("Validation Images:", len(val_dataset))
print("Testing Images  :", len(test_dataset))
```

```python
from collections import Counter
from utils.flower_names import FLOWER_NAMES

counter = Counter(train_dataset._labels)

for label, count in sorted(counter.items()):
    print(f"{FLOWER_NAMES[label]:30} : {count}")
```

```python
#View the number of categories
num_classes = len(set(train_dataset._labels))

print("Number of Classes:", num_classes)
```
